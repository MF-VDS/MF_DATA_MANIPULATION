### 🔴 Étape A – Extraction et étirement du NIR1.38 (Rouge)

La valeur maximale pour 10 % de réflectance est 706 (d'après `gdalinfo`).</br>
Option : -ot byte: spécifie le type de données de sortie en 8bits. Ici limitation à une plage de 0 à 255 </br>

```python
!gdalinfo -mm NETCDF:"$input":IR_013 | grep "Computed Min/Max" 2>/dev/null
```

```python
!gdal_translate -q -ot byte -scale 0 706 0 255 NETCDF:"$input":IR_013 {output}/NIR1.38_scale.tif
!gdalwarp -q -overwrite -ts 500 500 {output}/NIR1.38_scale.tif {output}/NIR1.38_scale.jpg
im2 = Image.open(output + '/NIR1.38_scale.jpg')
display(im2)
```

### 🟢 Étape B – Extraction du VIS0.64 (Vert) et NIR1.61 (Bleu)

Plage 0 → 80 % (valeur brute 8000).

```python
!gdal_translate -ot byte -scale 0 8000 0 255 NETCDF:"$input":VIS006 {output}/VIS0.64_scale.tif 2>/dev/null
!gdal_translate -ot byte -scale 0 8000 0 255 NETCDF:"$input":IR_016 {output}/NIR1.61_scale.tif 2>/dev/null
```

### 🔴 Étape C - Correction Gamma (avec `gdal_calc.py`)

La correction gamma modifie la luminosité non linéairement :

- **Gamma < 1** → assombrit l’image (augmente le contraste)
- **Gamma > 1** → éclaircit l’image (révèle les détails dans les ombres)

Formule : `out = max_in * ((in - min_in) / (max_in - min_in))^(1/gamma)`

###  Gamma NIR1.38 (γ = 1.5)

```python
file_in = 'RESULTS/NIR1.38_scale.tif'
imgpil = Image.open(file_in)
img = np.array(imgpil)
max_val = np.max(img)
min_val = np.min(img)
```

```python
gamma = 1.5
!gdal_calc.py --quiet --NoDataValue 0 --overwrite -A {file_in} --calc="numpy.maximum(A*0.,{max_val}*((A-{min_val})/({max_val}-{min_val}))**(1/{gamma}))" --outfile {output}/NIR1.38_scale_gamma.tif
!gdalwarp -q -overwrite -ts 500 500 {output}/NIR1.38_scale_gamma.tif {output}/NIR1.38_scale_gamma.jpg
im3 = Image.open(output + '/NIR1.38_scale_gamma.jpg')
display(im3)
```

### 🟢 Gamma VIS0.64 (γ = 0.75)

```python
file_in = 'RESULTS/VIS0.64_scale.tif'
imgpil = Image.open(file_in)
img = np.array(imgpil)
max_val = np.max(img)
min_val = np.min(img)
```

```python
gamma = 0.75
!gdal_calc.py --quiet --NoDataValue 0 --overwrite -A {file_in} --calc="numpy.maximum(A*0.,{max_val}*((A-{min_val})/({max_val}-{min_val}))**(1/{gamma}))" --outfile {output}/VIS0.64_scale_gamma.tif
!gdalwarp -overwrite -ts 500 500 {output}/VIS0.64_scale_gamma.tif {output}/VIS0.64_scale_gamma.jpg
im4 = Image.open(output + '/VIS0.64_scale_gamma.jpg')
display(im4)
```

### 🔵 Gamma NIR1.61 (γ = 1.0 – identité)

```python
file_in = 'RESULTS/NIR1.61_scale.tif'
imgpil = Image.open(file_in)
img = np.array(imgpil)
max_val = np.max(img)
min_val = np.min(img)
```

```python
gamma = 1.0
!gdal_calc.py --quiet --NoDataValue 0 --overwrite -A {file_in} --calc="numpy.maximum(A*0.,{max_val}*((A-{min_val})/({max_val}-{min_val}))**(1/{gamma}))" --outfile {output}/NIR1.61_scale_gamma.tif
!gdalwarp -overwrite -ts 500 500 {output}/NIR1.61_scale_gamma.tif {output}/NIR1.61_scale_gamma.jpg
im5 = Image.open(output + '/NIR1.61_scale_gamma.jpg')
display(im5)
```

## 🔴 Étape D - Fusion des couches (R, G, B)

`gdal_merge.py -separate` assemble trois fichiers en une image RVB.

</div>

```python
!gdal_merge.py -separate {output}/NIR1.38_scale_gamma.tif {output}/VIS0.64_scale_gamma.tif {output}/NIR1.61_scale_gamma.tif -o {output}/RGB_cloudtype.tif
!convert -resize 500 {output}/RGB_cloudtype.tif {output}/RGB_cloudtype_min.jpg > /dev/null 2>&1
im6 = Image.open(output + '/RGB_cloudtype_min.jpg')
display(im6)
```

<div class="alert alert-info alert-success">
<h3> 4️⃣ - 🗺️ Découpes et reprojections »</h3>
</div>


On peut extraire une sous‑zone (par exemple la région Afrique de l’Ouest) et la reprojeter dans différents systèmes.



### 🌍 Découpe en WGS84 (EPSG:4326)

```python
nord = 40
sud = 0
ouest = -23
est = 8
```

```python
!gdalwarp -q -t_srs "EPSG:4326" -te {ouest} {sud} {est} {nord} -overwrite {output}/RGB_cloudtype.tif {output}/RGB_cloudtype_zoom.tif
!convert -resize 500 {output}/RGB_cloudtype_zoom.tif {output}/RGB_cloudtype_zoom_min.jpg > /dev/null 2>&1
im7 = Image.open(output + '/RGB_cloudtype_zoom_min.jpg')
display(im7)
```

### 🗺️ Projection orthographique (vue satellite centrée sur la France)

```python
latitude = 45
longitude = 1
taille_domaine = "-1720000 -990000 1720000 990000"
```

```python
!gdalwarp -q -overwrite -t_srs "+proj=ortho lat_0={latitude} lon_0={longitude}" -te {taille_domaine} {output}/RGB_cloudtype.tif {output}/RGB_cloudtype_zoom_ortho.tif
!gdal_rasterize -b 1 -burn 250 -b 2 -burn 250 -b 3 -burn 250 -l world-administrative-boundaries OUTILS/boundary/world-administrative-boundaries.shp {output}/RGB_cloudtype_zoom_ortho.tif >/dev/null 2>&1
!convert -resize 800 {output}/RGB_cloudtype_zoom_ortho.tif {output}/RGB_cloudtype_zoom_ortho_min.jpg >/dev/null 2>&1
im8 = Image.open(output + '/RGB_cloudtype_zoom_ortho_min.jpg')
display(im8)
```

### 🧭 Projection Mercator (EPSG:3857)

Les coordonnées doivent être converties en mètres. On utilise `cs2cs`.

```python
!echo {ouest} {sud} | cs2cs +init=epsg:4326 +to +init=epsg:3857
!echo {est} {nord} | cs2cs +init=epsg:4326 +to +init=epsg:3857
```

```python
!gdalwarp -t_srs "EPSG:3857" -te -2560348.29 0 890555.93 4865942.28 -overwrite {output}/RGB_cloudtype_zoom.tif {output}/RGB_cloudtype_zoom_3857.tif
!convert -resize 500 {output}/RGB_cloudtype_zoom_3857.tif {output}/RGB_cloudtype_zoom_3857_min.jpg >/dev/null 2>&1
im9 = Image.open(output + '/RGB_cloudtype_zoom_3857_min.jpg')
display(im9)
```
