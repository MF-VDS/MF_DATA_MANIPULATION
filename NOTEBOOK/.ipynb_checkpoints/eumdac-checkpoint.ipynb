<b>Utilisation du EUMETSAT Data Access Client (EUMDAC)</b>

EUMDAC est le client d'accès aux données d'EUMETSAT. Il permet d'accéder aux données EUMETSAT de diverses missions satellitaires. En tant que bibliothèque Python, elle propose de nombreuses méthodes et aides pour utiliser les API et services d'EUMETSAT, comme Data Store et Data Tailor. Elle propose également divers utilitaires en ligne de commande pour la recherche, le téléchargement et la personnalisation des données.

### Avant la première utilisation, assurez-vous d'être enregistré auprès d'EUMETSAT.

Vous pouvez récupérer vos login et mot de passe dans vos Informations d'identification de l'utilisateur ("User credentials") via cette page https://api.eumetsat.int/api-key/#

### Utiliser vos identifiants via 2 possibilités :
- Créer un fichier .netrc sous votre home, en utilsant vos informations login et password que vous avez récupéré, contenant ces 3 lignes :</br>
machine api.eumetsat.int</br>
login <em>votre_login</em></br>
password <em>votre_mot_de_passe</em></br>
- Ou lancer cette commande : eumdac set-credentials <em>votre_login</em> <em>votre_mot_de_passe</em>

```python
eumdac set-credentials votre_login votre_mot_de_passe
```
